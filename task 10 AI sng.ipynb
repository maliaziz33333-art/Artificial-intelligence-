```python
# Question 5: Filling Null Values

# Upload file
from google.colab import files
uploaded = files.upload()

# Import library
import pandas as pd

# Load dataset
data = pd.read_csv('songs_db.csv')

# Null values before filling
print("Null Values Before Filling:\n")
print(data.isnull().sum())

# Fill object columns with mode
for col in data.select_dtypes(include='object').columns:

    mode_value = data[col].mode()[0]

    data[col] = data[col].fillna(mode_value)

# Fill numeric columns with mean
for col in data.select_dtypes(include=['int64', 'float64']).columns:

    mean_value = data[col].mean()

    data[col] = data[col].fillna(mean_value)

# Null values after filling
print("\nNull Values After Filling:\n")
print(data.isnull().sum())
```

```python
# Question 6: Type Conversion to Integer

# Upload file
from google.colab import files
uploaded = files.upload()

# Import libraries
import pandas as pd
import numpy as np

# Load dataset
data = pd.read_csv('songs_db.csv')

# Datatypes before conversion
print("Datatypes Before Conversion:\n")
print(data.dtypes)

# Convert float columns to integer
for col in data.select_dtypes(include=['float64']).columns:

    try:
        data[col] = data[col].astype(np.int64)
    except:
        pass

# Datatypes after conversion
print("\nDatatypes After Conversion:\n")
print(data.dtypes)
```

```python
# Question 10: Converting Object Columns into Integer Columns

# Upload file
from google.colab import files
uploaded = files.upload()

# Import library
import pandas as pd

# Load dataset
data = pd.read_csv('songs_db.csv')

# Split into X and Y
x = data.iloc[:, 0:-1]

# Find categorical columns
cat_columns = x.select_dtypes(['object']).columns

print("Categorical Columns:\n")
print(cat_columns)

# Convert object columns into integers
x[cat_columns] = x[cat_columns].apply(lambda col: pd.factorize(col)[0])

# Display converted dataset
print("\nConverted Dataset:\n")
print(x.head())

# Display datatypes
print("\nDatatypes After Conversion:\n")
print(x.dtypes)
```
